```python
import pandas as pd
import os
```

```python
# Construct a dataframe with the following columns:
# - batch size: {4, 8}
# - policy: {eager, latency-only, lazy, median, rebatching}
# - throughput
# - bert score

df = pd.DataFrame(columns=["batch_size", "policy", "throughput", "bert_score"])

for batch_size in [4, 8]:
    data_dir = f"./req_500_batch_{batch_size}/"
    for policy in ["eager", "latency-only", "lazy", "median", "rebatching"]:
        for file in os.listdir(data_dir):
            if file.endswith(".csv") and policy in file:
                file_path = os.path.join(data_dir, file)
                print(f"Processing {file_path}")
                try:
                    data = pd.read_csv(file_path)

                    throughput = data["throughput"].values[0]
                    bert_score = data["bert_score"].values[0]

                    # Add the data to the dataframe but dataframe object doe not has attribute append
                    df = pd.concat([df, pd.DataFrame({
                        "batch_size": [batch_size],
                        "policy": [policy],
                        "throughput": [throughput],
                        "bert_score": [bert_score]
                    })], ignore_index=True)
                except Exception as e:
                    print(f"Error processing {file_path}: {e}. Need to add manually.")

df.to_csv("multi_exits_70b.csv", index=False)


```
