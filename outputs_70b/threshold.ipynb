```python
import pandas as pd
from pathlib import Path
import re
import numpy as np
```

```python
data_dir = Path("./num_ee_threshold/batch_8")
```

```python
nums = [f"num_{i}" for i in range(6)]
throughputs = []
ee_props = []
involuntary_stay_lst= []
involuntary_ee_lst = []
for num in nums:
    dir = data_dir / num
    for csv_path in dir.glob("*.csv"):
        df = pd.read_csv(csv_path)

        throughputs.append(df["throughput"].mean())

        ee_prop = df["num_ee_tokens"].mean() / (df["num_ee_tokens"].mean() + df["num_no_ee_tokens"].mean()) * 100
        inv_stay = df["num_seq_would_ee_but_stay"].mean() / df["num_no_ee_tokens"].mean() * 100
        inv_ee = df["num_seq_would_not_ee_but_ee"].mean() / df["num_ee_tokens"].mean() * 100

        ee_props.append(ee_prop)
        involuntary_stay_lst.append(inv_stay)
        involuntary_ee_lst.append(inv_ee)

df = pd.DataFrame({
    "num": nums,
    "throughput": throughputs,
    "ee_prop": ee_props,
    "involuntary_stay": involuntary_stay_lst,
    "involuntary_ee": involuntary_ee_lst
})
df




        

```
