```python
import pandas as pd
```

```python
def get_throughput(experiment_dir: str):
    data_df = pd.read_csv(f"{experiment_dir}/replica_0/sequence_metrics.csv")
    throughput = data_df['request_num_tokens'].sum() / data_df['request_e2e_time'].sum()
    print(f"total tokens: {data_df['request_num_tokens'].sum()}")
    return throughput
```

```python
get_throughput("dataset_arxiv_model_llama-3-8b_tp_1_attn_fa_vattn_2mb_qps_1_reqs_1")
```
